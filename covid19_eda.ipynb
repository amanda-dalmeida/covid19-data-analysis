Vaticano — inconsistência populacional

O registro apresenta population(k) = 0.799, equivalente a aproximadamente 799 habitantes. Para verificar a plausibilidade desse valor, foram consultadas fontes externas sobre a população do Vaticano em 2026. Estimativas para 2026 indicam aproximadamente 506 habitantes, enquanto a área territorial oficial do Estado é de 0,44 km².

Embora o valor registrado não seja incompatível em ordem de grandeza com a população do Vaticano, ele não corresponde à referência populacional encontrada para o período analisado. Como a população é uma variável fundamental para análises normalizadas por população e o registro representa uma entidade com características demográficas extremamente particulares, optou-se por excluí-lo do dataset utilizado na análise
